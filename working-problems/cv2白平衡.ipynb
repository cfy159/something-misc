```python
import cv2

cap = cv2.VideoCapture(3)
while True:
    ret, frame = cap.read()
    if ret:
        cv2.imshow('frame', frame)

        if cv2.waitKey(1) & 0xFF == ord('s'):
            cv2.imwrite('white_balance_original.jpg', frame)
            break
        elif cv2.waitKey(1) & 0xFF == ord('q'):
            break
    else:
        break
        
cap.release()
cv2.destroyAllWindows()
```

```python

import cv2
import numpy as np

def simple_white_balance(img):
    wb = cv2.xphoto.createSimpleWB()
    wb.setInputMin(15)  # 优化暗区过滤阈值
    wb.setOutputMax(180)  # 防止高光过曝
    return wb.balanceWhite(img)

def grey_white_balance(img):
    wb = cv2.xphoto.createGrayworldWB()
    wb.setSaturationThreshold(10)  # 适应更多场景
    return wb.balanceWhite(img)

def load_and_balance(image_path):
    img = cv2.imread(image_path)
    if img is None:
        raise ValueError("图像加载失败，请检查路径")
    
    # 确保输入为uint8类型
    if img.dtype != np.uint8:
        img = cv2.convertScaleAbs(img)
    
    gray_balanced_img = grey_white_balance(img)  # 或切换为simple_white_balance
    simple_balance_img = simple_white_balance(img)

    
    cv2.imshow("Original", img)
    cv2.imshow("gray", gray_balanced_img)
    cv2.imshow("simple", simple_balance_img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()


load_and_balance("white_balance_original.jpg")
```
